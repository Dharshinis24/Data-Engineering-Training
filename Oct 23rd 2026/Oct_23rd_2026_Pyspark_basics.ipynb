**Joins**

```python
customers_data = [
    ("C00001", "Aman Retail", "Hyderabad", "Corporate"),
    ("C00002", "Blue Mart", "Mumbai", "Retail"),
    ("C00003", "Fresh Basket", "Bangalore", "Retail"),
    ("C00004", "Prime Stores", "Delhi", "SMB"),
    ("C00005", "Quick Shop", "Pune", "Corporate")
]

customers_columns = [
    "customer_id",
    "customer_name_master",
    "customer_city",
    "segment"
]

customers_df = spark.createDataFrame(
    customers_data,
    customers_columns
)

customers_df.show()
```

```python
joined = df.join(
    customers_df,
    on='customer_id',
    how='inner'
)
joined.show()
```

```python
left_join = df.join(
    customers_df,
    on='customer_id',
    how='left'
)
left_join.show()
```

```python
right_join = df.join(
    customers_df,
    on='customer_id',
    how='right'
)
right_join.show()
```

```python
full_join = df.join(
    customers_df,
    on='customer_id',
    how='full'
)
full_join.show()
```

**Working with window function**

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import (
    col,
    row_number,
    rank,
    dense_rank,
    sum,
    avg,
    lag,
    lead
)
```

```python
# Grouping by category and product_name to compute the average price
product_df = df.groupBy(
    'category',
    'product_name'
).agg(
    avg('unit_price').alias('avg_price')
)
product_df.show()
```

```python
window_spec = Window.partitionBy('category').orderBy(col('avg_price').desc())

ranked_products = product_df.withColumn(
    'rank',
    row_number().over(window_spec)
)
ranked_products.show()
```

**Working with date**

```python
from pyspark.sql.functions import col, to_date,year,month,dayofmonth
df2 = df.withColumn(
    'order_date',
    to_date(col('order_date'),'yyyy-MM-dd')
)
df2.show()
```

```python
df2 = df2.withColumn(
    'year',
    year(col('order_date'))
)
df2.select(
    'order_date',
    'year'
).show()
```

```python
df2 = df2.withColumn(
    'month',
    month('order_date')
)
df2.select(
    'order_date',
    'month'
).show()
```

```python
df2 = df2.withColumn(
    'date',
    dayofmonth('order_date')
).select(
    'order_date',
    'date'
).show()
```
