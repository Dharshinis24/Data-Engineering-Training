Step 1 — Create the Delta table

```python
%sql
CREATE OR REPLACE TABLE fitness_memberships_demo (
member_id INT,
member_name STRING,
city STRING,
membership_plan STRING,
monthly_fee INT,
duration_months INT,
status STRING
)
USING DELTA;
```
