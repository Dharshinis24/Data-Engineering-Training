Part C — MERGE and UPSERT (SCD Type 1)

```python
%sql
CREATE OR REPLACE TEMP VIEW incoming_parcels AS
SELECT *
FROM VALUES
(102, 'Meera', 'Mumbai', 'Express', 4.0, 650, 'Delivered'),
(105, 'Rohit', 'Chennai', 'Express', 5.0, 900, 'Delivered'),
(117, 'Ayesha', 'Kolkata', 'Standard', 2.0, 300, 'Pending'),

(118, 'Rahul', 'Jaipur', 'Express', 3.0, 600, 'In Transit')
AS incoming(
parcel_id,
customer_name,
destination_city,
service_type,
weight_kg,
shipping_fee,
delivery_status
);
```

```python
%sql
--11
select *
from incoming_parcels;
```

```python
%sql
--12
select *
from courier_parcels_delta cpd
join incoming_parcels ip
on cpd.parcel_id = ip.parcel_id;
```

```python
%sql
--13
select *
from incoming_parcels ip   
left join courier_parcels_delta cpd 
on ip.parcel_id = cpd.parcel_id
where cpd.parcel_id is null;
```

```python
%sql
--14 
merge into courier_parcels_delta cpd
using incoming_parcels ip    
on cpd.parcel_id = ip.parcel_id

when matched then 
update set 
    cpd.customer_name = ip.customer_name,
    cpd.destination_city = ip.destination_city,
    cpd.service_type = ip.service_type,
    cpd.weight_kg = ip.weight_kg,
    cpd.shipping_fee = ip.shipping_fee,
    cpd.delivery_status = ip.delivery_status

when not matched then 
insert (parcel_id, customer_name, destination_city, service_type, weight_kg, shipping_fee, delivery_status)
values (ip.parcel_id, ip.customer_name, ip.destination_city, ip.service_type, ip.weight_kg, ip.shipping_fee, ip.delivery_status);
```
