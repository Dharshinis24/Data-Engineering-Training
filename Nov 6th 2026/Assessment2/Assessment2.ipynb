Section C — PySpark DataFrames

```python
#13
event_registration_df = spark.sql("select * from event_registrations_source")

event_master_df = spark.sql("select * from event_master")

event_registration_df.printSchema()
event_master_df.printSchema()
```

```python
#14
event_registration_df.select(
    'registration_id',
    'attendee_name',
    'tickets',
    'unit_price'
).show()
```

```python
#15
from pyspark.sql.functions import col
event_registration_df.filter(
    (col('status') == 'Confirmed') &
    (col('tickets') > 2)
).show()
```

```python
#16
event_registration_df = event_registration_df.withColumn(
    'total_amount',
    col('tickets') * col('unit_price')
)
event_registration_df.show()
```

```python
#17
from pyspark.sql.functions import when 
event_registration_df = event_registration_df.withColumn(
    'booking_category',
    when(col('total_amount') >= 2000,'High')
    .otherwise('Low')
)
event_registration_df.show()
```

```python
#18
from pyspark.sql.functions import sum
joined_df = event_registration_df.join(
    event_master_df,
    on = 'event_id',
    how = 'inner'
).filter(
    col('status') == 'Confirmed'
).groupBy(col('city')).agg(
    sum(col('total_amount')).alias('total_confirmed_ticket_revenue')
)
joined_df.show()
```

```python
%sql
--26
merge into event_registrations_work erm
using incoming_registrations ir 
on erm.registration_id = ir.registration_id 
when matched then 
update set 
    erm.event_id = ir.event_id,
    erm.attendee_name = ir.attendee_name,
    erm.registration_date = ir.registration_date,
    erm.ticket_type = ir.ticket_type,
    erm.tickets = ir.tickets,
    erm.unit_price = ir.unit_price,
    erm.status = ir.status,
    erm.channel = ir.channel
when not matched then
insert (registration_id, event_id, attendee_name, registration_date, ticket_type, tickets, unit_price, status, channel)
values (ir.registration_id, ir.event_id, ir.attendee_name,ir.registration_date, ir.ticket_type, ir.tickets, ir.unit_price, ir.status, ir.channel)
```

Section I — Structured Streaming

```python
import time
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("Spark DataFrames").getOrCreate()

streaming_df = spark.readStream.format('rate').option('rowsPerSecond',5).load()

query = (streaming_df.writeStream
    .format('console')
    .outputMode('append')
    .option('checkpointLocation', '/Volumes/workspace/default/demo_volume/section_i_checkpoint')
    .trigger(availableNow=True)
    .start())

try:
    time.sleep(15)
finally:
    query.stop()
```

```python
#47
# streaming DataFrames cannot be executed with .show() -- drain a bounded batch to a memory table and display it
# unique checkpoint folder per run: the rate source cannot recover from an existing checkpoint
query = (streaming_df.writeStream
    .format('memory')
    .queryName('rate_stream_output')
    .outputMode('append')
    .option('checkpointLocation', f'/Volumes/workspace/default/demo_volume/section_i_display_{time.time_ns()}')
    .trigger(availableNow=True)
    .start())
query.awaitTermination()
display(spark.table('rate_stream_output'))
```

```python
from pyspark.sql.functions import col
streaming_df = streaming_df.withColumn(
    'transaction_id',
    col('value') + 1000
).withColumn(
    'transaction_amount',
    col('value')*10
)
display(
    streaming_df,
    checkpointLocation=f'/Volumes/workspace/default/demo_volume/section_i_transformed_{time.time_ns()}'
)
```

```python
#49
# streaming DataFrames cannot be executed with .show() -- display() renders the live streaming view
# unique checkpoint folder per run: the rate source cannot recover from an existing checkpoint
display(
    streaming_df.filter(col('transaction_amount') > 500),
    checkpointLocation=f'/Volumes/workspace/default/demo_volume/section_i_filter_{time.time_ns()}'
)
```
